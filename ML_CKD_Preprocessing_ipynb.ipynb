**Recursive Feature Elimination**

The scaling you did during KNN imputation was applied to individual columns or small groups of columns to help the imputer. However, for RFE, you typically want to scale all your independent variables (indep_X) together after all the imputation and preprocessing steps are complete and after you've split your data into training and testing sets.

Scaling the entire feature set before applying RFE ensures that features with different ranges of values don't unfairly influence the feature selection process based on their magnitude. So, yes, you should still apply StandardScaler to your training and testing sets (using a scaler fitted only on the training data) before using RFE.

```python
# Define the split and scalar function (assuming it's not already in the notebook)
def split_scalar(indep_X,dep_Y):
        from sklearn.model_selection import train_test_split
        from sklearn.preprocessing import StandardScaler

        X_train, X_test, y_train, y_test = train_test_split(indep_X, dep_Y, test_size = 0.25, random_state = 0)

        sc = StandardScaler()
        X_train = sc.fit_transform(X_train)
        X_test = sc.transform(X_test)
        return X_train, X_test, y_train, y_test

# Define the classification evaluation function to include precision, recall, and F1-score
def evaluate_classification_model(regressor, X_test, y_test):
    from sklearn.metrics import accuracy_score
    y_pred = regressor.predict(X_test)

    accuracy = accuracy_score(y_test, y_pred)
    # Precision, recall, and F1-score removed as per user request
    # precision = precision_score(y_test, y_pred, average='binary', pos_label=1)
    # recall = recall_score(y_test, y_pred, average='binary', pos_label=1)
    # f1 = f1_score(y_test, y_pred, average='binary', pos_label=1)

    return accuracy # Only return accuracy


# Define the rfeFeature function to use a single specified model
def rfeFeature_single_model(indep_X, dep_Y, n, model, model_name):
    """
    Performs RFE with a single specified model.

    Args:
        indep_X (pd.DataFrame): Independent variables.
        dep_Y (pd.DataFrame): Dependent variable (target).
        n (int): Number of features to select.
        model: The scikit-learn model to use for RFE.
        model_name (str): The name of the model for printing.

    Returns:
        tuple: (rfe_feature_train, rfe_feature_test, selected_columns, evaluation_metrics)
               Contains the RFE transformed training and testing data,
               the list of selected column names, and a tuple of evaluation metrics (accuracy, precision, recall, f1).
    """
    print(f"Running RFE with {model_name}...")

    # Split and scale data
    X_train, X_test, y_train, y_test = split_scalar(indep_X, dep_Y)

    # RFE fitting needs the original scaled training data
    from sklearn.feature_selection import RFE
    log_rfe = RFE(estimator=model, n_features_to_select=n)
    # RFE fit expects 1D array for y for classification
    log_fit = log_rfe.fit(X_train, y_train.values.ravel()) # .values.ravel() converts DataFrame to 1D array
    rfe_feature_train = log_fit.transform(X_train)
    rfe_feature_test = log_fit.transform(X_test)

    # Get the column names selected by RFE based on the original columns
    selected_columns = [col for col, selected in zip(indep_X.columns, log_rfe.support_) if selected]

    # Fit the model on the RFE selected features and calculate evaluation metric
    model.fit(rfe_feature_train, y_train.values.ravel()) # Fit on RFE selected training data
    evaluation_value = evaluate_classification_model(model, rfe_feature_test, y_test) # Now returns only accuracy

    return rfe_feature_train, rfe_test_data, selected_columns, evaluation_value # Return only accuracy


# --- Example Usage with Logistic Regression ---
from sklearn.linear_model import LogisticRegression

# Instantiate the chosen model
logistic_regression_model = LogisticRegression(solver='liblinear', random_state=0) # Added random_state for reproducibility
chosen_model_name = "Logistic Regression"
n_features_to_select = 5 # Specify the number of features you want to select


# Call the function with your data and the chosen model
# Update variable names to match the single return value from evaluate_classification_model
rfe_train_data, rfe_test_data, selected_cols, accuracy = rfeFeature_single_model(
    indep_X, dep_Y, n_features_to_select, logistic_regression_model, chosen_model_name
)

# Print the results
print(f"\n--- RFE Results for {chosen_model_name} ---")
print("Selected Columns:", selected_cols)
print(f"Accuracy on test set with selected features: {accuracy:.4f}\n")


# You can now use rfe_train_data_lr and rfe_test_data_lr for further model training and evaluation
```
