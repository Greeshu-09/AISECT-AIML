*Recursive Feature Elimination*

The scaling you did during KNN imputation was applied to individual columns or small groups of columns to help the imputer. However, for RFE, you typically want to scale all your independent variables (indep_X) together after all the imputation and preprocessing steps are complete and after you've split your data into training and testing sets.

Scaling the entire feature set before applying RFE ensures that features with different ranges of values don't unfairly influence the feature selection process based on their magnitude. So, yes, you should still apply StandardScaler to your training and testing sets (using a scaler fitted only on the training data) before using RFE.

```python
def split_scalar(X,Y):
        X_train, X_test, y_train, y_test = train_test_split(indep_X, dep_Y, test_size = 0.25, random_state = 0)

#creating function from classification metrics as a performance metric
def classification_metrics(classifier, X_test, y_test):
    # Predictions
    y_pred = classifier.predict(X_test)

    # Import metrics
    from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report

    # Calculate metrics
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, average='weighted')  # use 'binary' for binary problems
    recall = recall_score(y_test, y_pred, average='weighted')
    f1 = f1_score(y_test, y_pred, average='weighted')
    cm = confusion_matrix(y_test, y_pred)

    # Optional: classification report
    report = classification_report(y_test, y_pred)

    return {
        'accuracy': accuracy,
        'precision': precision,
        'recall': recall,
        'f1_score': f1,
        'confusion_matrix': cm,
        'classification_report': report
    }
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

def train_and_evaluate_logistic_regression(X, y):
    """
    Trains a Logistic Regression classifier, evaluates its performance, and returns metrics.

    Args:
        X: Features (DataFrame).
        y: Target (Series or DataFrame).

    Returns:
        A dictionary containing accuracy, precision, recall, f1_score, and confusion_matrix.
    """
    # Split data into training and testing sets
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)

    # Initialize and train the Logistic Regression classifier
    lr_classifier = LogisticRegression(random_state=42, solver='liblinear') # Added solver for clarity
    lr_classifier.fit(X_train, y_train)

    # Evaluate the classifier using the existing classification_metrics function
    metrics = classification_metrics(lr_classifier, X_test, y_test)

    return metrics
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split

def train_and_evaluate_random_forest(X, y):
    """
    Trains a Random Forest classifier, evaluates its performance, and returns metrics.

    Args:
        X: Features (DataFrame).
        y: Target (Series or DataFrame).

    Returns:
        A dictionary containing accuracy, precision, recall, f1_score, and confusion_matrix.
    """
    # Split data into training and testing sets
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)

    # Initialize and train the Random Forest classifier
    rf_classifier = RandomForestClassifier(random_state=42)
    rf_classifier.fit(X_train, y_train)

    # Evaluate the classifier using the existing classification_metrics function
    metrics = classification_metrics(rf_classifier, X_test, y_test)

    return metrics
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split

def train_and_evaluate_decision_tree(X, y):
    """
    Trains a Decision Tree classifier, evaluates its performance, and returns metrics.

    Args:
        X: Features (DataFrame).
        y: Target (Series or DataFrame).

    Returns:
        A dictionary containing accuracy, precision, recall, f1_score, and confusion_matrix.
    """
    # Split data into training and testing sets
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)

    # Initialize and train the Decision Tree classifier
    dt_classifier = DecisionTreeClassifier(random_state=42)
    dt_classifier.fit(X_train, y_train)

    # Evaluate the classifier using the existing classification_metrics function
    metrics = classification_metrics(dt_classifier, X_test, y_test)

    return metrics
```

```python
#this function automates the process of applying RFE with different classifiers, training those classifiers on the RFE-selected features, and evaluating their performance.
from sklearn.feature_selection import RFE
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

from sklearn.model_selection import train_test_split # Ensure train_test_split is imported here

def evaluate_models_with_rfe(X, y, n):
    """
    Applies RFE with various classifiers to select top features and evaluates models.

    Args:
        X: Features (DataFrame).
        y: Target (Series or DataFrame).
        n: The number of features to select.

    Returns:
        A dictionary where keys are model names and values are a dictionary containing their performance metrics and selected features.
    """
    results = {}

    # Define classifiers to use with RFE
    classifiers = {
        "Decision Tree": DecisionTreeClassifier(random_state=42),
        "Logistic Regression": LogisticRegression(random_state=42, solver='liblinear'),
        "Random Forest": RandomForestClassifier(random_state=42),
    }

    # Define classifiers to evaluate without RFE (if needed, or handle separately)
    # For now, focusing on RFE compatible models as per the function's purpose

    for name, classifier in classifiers.items():
        print(f"Evaluating {name} with RFE...")

        # Initialize RFE
        rfe = RFE(estimator=classifier, n_features_to_select=n)

        # Fit RFE
        rfe.fit(X, y)

        # Get the selected features
        selected_features = X.columns[rfe.support_].tolist()

        # Split data into training and testing sets using selected features
        X_selected = X[selected_features]
        X_train, X_test, y_train, y_test = train_test_split(X_selected, y, test_size=0.3, random_state=0)


        # Train the classifier on the selected features
        model = classifier
        model.fit(X_train, y_train)

        # Evaluate the classifier
        metrics = classification_metrics(model, X_test, y_test)
        results[name] = {'metrics': metrics, 'selected_features': selected_features}
        print(f"{name} Evaluation Complete.\n")

    return results
```

```python
n_features = 5

# Call the function with your data
rfe_results = evaluate_models_with_rfe(indep_X, dep_Y.values.ravel(), n_features)

# Print the results for each model in the requested format
for model_name, result_data in rfe_results.items():
    print(f"Model: {model_name}")
    print("Selected Columns:", result_data['selected_features'])
    # Print classification metrics
    print(f"Accuracy: {result_data['metrics']['accuracy']:.4f}")
    print(f"Precision: {result_data['metrics']['precision']:.4f}")
    print(f"Recall: {result_data['metrics']['recall']:.4f}")
    print(f"F1-Score: {result_data['metrics']['f1_score']:.4f}")
    print(f"Confusion Matrix:\n{result_data['metrics']['confusion_matrix']}\n")
```
